```python
# Import the Python SDK
import google.genai as genai # Changed from google.generativeai
# Used to securely store your API key
from google.colab import userdata

# Configure the generative AI library with your API key from Colab Secrets
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
genai.configure(api_key=GOOGLE_API_KEY)
```

Once configured, you can initialize a generative model. We'll use a suitable model for generating travel recommendations.

```python
# Initialize the Gemini Generative Model
gemini_model = genai.GenerativeModel('gemini-pro') # 'gemini-pro' is a good general-purpose model
```

Now that the model is initialized, you can use it to generate content. For example, let's ask it for a travel recommendation:

```python
response = gemini_model.generate_content('Suggest a 3-day itinerary for a relaxing trip to a beach destination.')
print(response.text)
```
